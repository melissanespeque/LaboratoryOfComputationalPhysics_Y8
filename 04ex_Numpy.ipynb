7\. Prime numbers sieve: compute the prime numbers in the 0-N (N=99 to start with) range with a sieve (mask).
  * Constract a shape (100,) boolean array, the mask
  * Identify the multiples of each number starting from 2 and set accordingly the corresponding mask element
  * Apply the mask to obtain an array of ordered prime numbers
  * Check the performances (timeit); how does it scale with N?
  * Implement the optimization suggested in the [sieve of Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes)

````python
N = 500
mask = np.ones(N,dtype = bool) #isso cria um array boolean de shape 100, com trues(1), pois é uma mask
mask[0] = False #0 nao é numero primo
mask[1] = False #1 não é numero primo

for i in range(2, N):
     # marca múltiplos de i
    mask[2*i::i] = False
primes = np.arange(N)[mask]

print(primes)
````

Otimização do Sieve de Eratóstenes (não precisa avaliar TODOS os numeros)

A otimização clássica é:

(A) parar em √N: Porque se um número tem divisor, um deles necessariamente é ≤ √N. \
(B) começar em i² Porque múltiplos menores já foram eliminados antes.

````python
import timeit

def sieve_basic(N):
    mask = np.ones(N, dtype=bool)
    mask[:2] = False
    for i in range(2, int(np.sqrt(N))+1):
        if mask[i]:
            mask[i*i::i] = False
    return np.arange(N)[mask]

for N in [10_000, 100_000, 1_000_000]:
    t = timeit.timeit(lambda: sieve_basic(N), number=3)
    print('for N iterations', N, 'Performance time', t)
````

8\. Diffusion using random walk

Consider a simple random walk process: at each step in time, a walker jumps right or left (+1 or -1) with equal probability. The goal is to find the typical distance from the origin of a random walker after a given amount of time. 
To do that, let's simulate many walkers and create a 2D array with each walker as a raw and the actual time evolution as columns

  * Take 1000 walkers and let them walk for 200 steps
  * Use randint to create a 2D array of size walkers x steps with values -1 or 1
  * Build the actual walking distances for each walker (i.e. another 2D array "summing on each raw")
  * Take the square of that 2D array (elementwise)
  * Compute the mean of the squared distances at each step (i.e. the mean along the columns)
  * Plot the average distances (sqrt(distance\*\*2)) as a function of time (step)
  
Did you get what you expected?

````python
walker = 1000
step = 200
#np.random.randint(0, 2, size=(walkers, steps)) gera randons inteiros entre 0 2, ou seja, 0 ou 1
#depois para transformar em -1 e 1: faz numero*2 - 1 -> 0 vai pra -1 e 1 vai pra 1
moves = 2*(np.random.randint(0,2,size=(walker,step)))-1
#cada linha, um walker difereente
moves
````

````python
positions = np.cumsum(moves, axis=1) #faço a soma acumulada, axis=1 para somar HORIZONTALMENTE (CADA LINHA, UM WALKER) 
positions
positions = positions**2
positions
````

````python
mean_squared = np.mean(positions, axis=0) #mean along the columns: olhada VERTICALMENTE
mean_squared
rms_distance = np.sqrt(mean_squared) #em media onde cada walker está em cada passo? por isso gera um grafico de media. para obter a difusão de random walker
rms_distance
````

````python
plt.plot(rms_distance)

plt.xlabel("Step")
plt.ylabel("Typical Distance")

plt.title("Random Walk Diffusion")

plt.show()
````
