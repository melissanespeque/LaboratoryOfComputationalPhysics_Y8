```python
#ALASKA TEMPERATURE
import numpy as np

maxt =np.array([ 17, 19, 21, 28, 33, 38, 37, 37, 31, 23, 19, 18])
mint = np.array([-62, -59, -56, -46, -32, -18, -9, -13, -25, -46, -52, -58])
month = np.arange(1,13) 
```

```python
plt.plot(month,maxt,'r-o')
plt.plot(month,mint,'b-o')
```

```python
def temperature(month,amp,offset,avg):
    temp = amp*np.sin(2*np.pi*(month-offset)/12) + avg
    return temp

from scipy.optimize import curve_fit
```

```python
p0 = [40,4,35]

paramax, _ = curve_fit(temperature,month,maxt,p0)
paramin, _ = curve_fit(temperature,month,mint)

print(paramax)
```

```python
plt.plot(month,maxt,'ro')
plt.plot(month,temperature(month,*paramax),'r-')
plt.plot(month,mint,'bo')
plt.plot(month,temperature(month,*paramin),'b-')
plt.grid(True)
```
