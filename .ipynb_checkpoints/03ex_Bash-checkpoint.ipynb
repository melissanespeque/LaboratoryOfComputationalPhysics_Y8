2\.c Distinguish the entries on the basis of `sqrt(X^2 + Y^2 + Z^2)` is greater or smaller than `100*sqrt(3)/2`. Count the entries of each of the two groups 

calcular sqrt(x^2 + y^2 + z^2) result = g
ver se g é gt or lt 100 * sqrt(3) / 2
contar quantos são mais e quantos são 
código:

awk '{r=sqrt($1^2 + $2^2 + $3^2)

    limit=100*sqrt(3)/2
    
    if (r > limit) high++
    
    else low++}
    
    END{
    
        print "smaller or equal:",low
        
        print "greater:",high}' data.txt
        
smaller or equal: 12
greater: 38

limit=$(echo "100*sqrt(3)/2" | bc -l)

high=0
low=0

while read -r x y z rest; do

    r=$(echo "sqrt($x^2 + $y^2 + $z^2)" | bc -l)

    if (( $(echo "$r > $limit" | bc -l) )); then
        high=$((high + 1))
    else
        low=$((low + 1))
    fi
done < data.txt

echo "Smaller or equal: $low"

echo "Greater: $high"


2\.d Make `n` copies of data.txt (with `n` an input parameter of the script), where the i-th copy has all the numbers divided by i (with `1<=i<=n`).

n=$1

for ((i=1; i<=n; i++)); do
    > data_$i.txt   # cria/limpa arquivo

    while read -r line; do
        newline=""
        for num in $line; do
            value=$(echo "scale=6; $num/$i" | bc -l)
            newline="$newline$value "
        done
        echo "$newline" >> data_$i.txt
    done < data.txt
done
