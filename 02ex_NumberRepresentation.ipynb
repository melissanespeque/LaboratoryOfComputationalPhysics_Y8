1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
#começando do bin para dec ex: 1111 = 15
def bintodec(bin):
    size = len(bin)
    dec = sum(int(bin[i]) * 2**(size - 1 - i) for i in range(size)) #itera a soma em cada posição-2^0 até 2^size-1
    return dec

def dectobin(dec):
    rest = []
    while dec > 0:
        r = dec%2
        rest.append(r)
        dec = dec//2 #/ = divisão float; // = divisão inteira
        rest.reverse()
        binary_str = "".join(str(b) for b in rest)
    return binary_str

def hextodec(hex):
#de 0 a 9 = 0 a 9
#dicionario de A a F
#posição: 16**p X numero -> 2A = 2x16**1 + 10x16**0 = 32 + 10 = 42
    hex_map = {'A': 10, 'B': 11, 'C': 12, 'D': 13, 'E': 14, 'F': 15}
    size = len(hex)
    dec = 0
    for i in range(size):
        digit = hex[i]
        # se for número 0–9
        if digit.isdigit():
            value = int(digit)
        else:
            value = hex_map[digit]
        power = size - 1 - i
        dec += value * (16 ** power)

    return dec

def dectohex(dec):
    hex_map = {10:'A', 11:'B', 12:'C', 13:'D', 14:'E', 15:'F'}
    digits = []
    while dec > 0:
        r = dec % 16
        if r < 10:
            digits.append(str(r))
        else:
            digits.append(hex_map[r])
        dec //= 16
    digits.reverse()
    return "".join(digits)
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
#escrever cada parte
#computa e junta td
#(-1)^s x (1+m) x 2^e
def floatconv(bits):
    s = int(bits[0]) #sign: first bit
    e = bits[1:9] #represetna os 8 proximos bits
    m = bits[9:] #os ultimos 23 bits
    exponent = int(e,2)-127 #int(valor,base) -> valor e na base 2(bin)
#fração binaria: 0.1 = 1/2;  0.01 = 1/4;  0.001 = 1/8;   0.0001 = 1/16  
    mantissa = 1.0 
    for i, bit in enumerate(m):
        if bit=='1':
            mantissa += 2**(-(i+1))
    value = ((-1)**s)*mantissa*(2**exponent)
    return value
floatconv("01000000101000000000000000000000") 
```
